```python
#Stwórz losową 1-wymiarową numpy array o długości 15 elementów. Następnie oblicz jej minimum, maksimum oraz odchylenie standardowe.
list1 = np.random.randint(100, size=15)
print(list1)

print(min(list1))
print(max(list1))
print(np.std(list1))
```

```python
# Stwórz losową tablicę 2-wymiarową numpy array 4 wiersze x 4 kolumny, a następnie oblicz sumę wartości dla każdej kolumny oraz dla każdego wiersza.
array1 = np.random.randint(1, 20, size=(4, 4))
print(array1)


print(f"W1 = {sum(array1[0])}")
print(f"W2 = {sum(array1[1])}")
print(f"W3 = {sum(array1[2])}")
print(f"W4 = {sum(array1[3])}")

print(f"K1 = {sum(array1[:,0])}")
print(f"K2 = {sum(array1[:,1])}")
print(f"K3 = {sum(array1[:,2])}")
print(f"K4 = {sum(array1[:,3])}")
```
