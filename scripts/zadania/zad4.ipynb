A. Dla warstwy z pasami ukształtowania policz ich powierzchnie w m2, korzystając z układu epsg:2180

```python
import geopandas as gpd
path_to_files = "../dane/"

zones = gpd.read_file(path_to_files+"pl_pasy_ukształtowania.geojson")
print(zones.crs)
print(zones.crs.name)

zones2 = zones.to_crs(epsg=2180)
print(zones2.crs)
print(zones2.crs.name)

print(zones2.iloc[0].geometry.area)
```
