A. Dla warstwy z pasami ukształtowania policz ich powierzchnie w m2, korzystając z układu epsg:2180

```python
import geopandas as gpd
path_to_files = "../dane/"

zones = gpd.read_file(path_to_files+"pl_pasy_ukształtowania.geojson")
print(zones.crs)
print(zones.crs.name)

zones2 = zones.to_crs(epsg=2180)
print(zones2.crs)
print(zones2.crs.name)

print(zones2.iloc[0].geometry.area)
```

B. Stwórz obiekt Bounding Box dla warstwy z miastami wojewódzkimi, a następnie oblicz jego obwód

```python
from shapely.geometry import box

miasta = gpd.read_file(path_to_files+"pl_miasta_wojewódzkie.geojson")
print(miasta.crs)

miasta = miasta.to_crs(epsg=2180)
print(miasta.crs)

miasta_bb = miasta.total_bounds
print(miasta_bb)

xmin, ymin, xmax, ymax = miasta_bb
miasta_bb_gdf = gpd.GeoDataFrame(
    {"name": ["miasta_bb"]},
    geometry=[box(xmin, ymin, xmax, ymax)],
    crs=miasta.crs,
)
print(miasta_bb_gdf.iloc[0].geometry.length)
```

C. Dla warstwy z rzekami stwórz nową kolumnę w której wyliczysz liczbę linii (LineString) znajdujących się wewnątrz. Analogicznie stwórz następnie drugą kolumnę w której wyliczysz łączną liczbę punktów składających się na geometrię

```python
rzeki = gpd.read_file(path_to_files+"pl_rzeki.geojson")

rzeki["liczba_lini"] = rzeki.geometry.apply(
    lambda geom: len(geom.geoms))
rzeki["liczba_punktow"] = rzeki.geometry.apply(
    lambda geom: sum(1 for line in geom.geoms for _ in line.coords))
```
