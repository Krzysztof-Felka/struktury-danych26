1. Utwórz DataFrame zawierający dane 10 osób, wzrost w centymetrach oraz masę ciała w kilogramach. Następnie na podstawie tych danych wygeneruj kolumnę zawierającą BMI każdej osoby.

```python
names = ["Ania", "Paweł", "Joachim", "Hanna", "Elżbieta", "Benjamin", "Ernest", "Paulina", "Władysław", "Żaneta"]
height = np.random.randint(150, 200, size=10)
height.astype(float)
weight = np.random.randint(60, 120, size=10)

dane=list(zip(names, height, weight))

df1 = pd.DataFrame(dane, columns=['names', 'height', 'weight'])
df1
df1["BMI"] = round(df1["weight"]/pow(df1["height"]/100, 2), 2)
df1
```

2. Utwórz własny DataFrame zawierający wybrane przez siebie dane. Zapisz go do pliku JSON, a następnie wczytaj ten plik do nowego obiektu DataFrame. Sprawdź, czy oba obiekty zawierają te same dane.

```python
df2 = pd.DataFrame({
    'produkt': ['jabłko', 'banan', 'gruszka'],
    'cena': [2.50, 3.20, 4.10],
    'liczba_sztuk': [10, 7, 5],
    'kategoria': ['owoce', 'owoce', 'owoce']
})
df2

df2.to_json('df2.json', orient='records', lines=True)

df2_j = pd.read_json('df2.json', orient='records', lines=True)
df2_j

df2 != df2_j

print(df2.equals(df2_j))
```

3. Wykorzystaj DataFrame stworzony w punkcie 1. Dodaj kolumnę z wiekiem i wypełnij ją danymi. Następnie znajdź osoby starsze niż średnia wieku wszystkich osób, których BMI jest jednocześnie większe niż 25. Wyświetl imiona, wiek oraz BMI tych osób.

```python
df1["age"] = np.random.randint(18, 65, size=10)
df1

wynik = df1[
    (df1["age"] > df1["age"].mean()) &
    (df1["BMI"] > 25)
]

wynik[["names", "age", "BMI"]]
```
