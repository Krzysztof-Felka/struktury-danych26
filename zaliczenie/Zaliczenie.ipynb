```python
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
import pandas as pd
import os
from pathlib import Path
import statistics as sts
```

```python
df = pd.DataFrame(columns=['nazwa_warstwy',
                            'rodzaj_geometrii',
                            'liczba_obiektów',
                              'srd_powierzchnia',
                                'srd_dlugosc',
                                  'srd_zlozonosc_lini',
                                    'srd_wypelnienie_przestrzeni'])
```

```python
path_shp = "dane/PL.PZGiK.308.BDOT10k.3012__OT_"

p = Path("dane/")
shp_files = [f for f in os.listdir('dane/') if f.endswith('.shp')]
print(*shp_files)

#for f in p.iterdir():
    #if f.is_file():
    #layer_name = path_shp + str(i).zfill(2) + ".shp"
    #layer = gpd.read_file(layer_name)
    
```

```python
lista_gpd = []
obj_nm_l = []

for i in shp_files:
    obj_nm = i[30:36]
    obj_nm_l.append(obj_nm)
    gpd_obj = gpd.read_file(path_shp + obj_nm + ".shp")
    lista_gpd.append(gpd_obj)
    
df['nazwa_warstwy'] = shp_files
```

```python
#adja = lista_gpd[0]
length_l = []
geom_count = []
geom_type_l = []
area_l = []

for i in lista_gpd:
    typ = i.geom_type.unique()
    geom_type_l.append(typ)
    count = sum(i.count_geometries())
    geom_count.append(count)
    if i.geom_type.unique() == ["Polygon"]:
        area = sts.mean(i.geometry.area)
        area_l.append(area)
    else:
        area_l.append(0)
    if i.geom_type.unique() == ["Polygon"]:
        length = sts.mean(i.geometry.length)
        length_l.append(length)
    elif i.geom_type.unique() == ["LineString"]:
        length = sts.mean(i.geometry.length)
        length_l.append(length)
    else: length_l.append(0)

# nie skończyłem w kolumn średniej złożoności i wypełnienia

df['liczba_obiektów'] = geom_count
df['rodzaj_geometrii'] = geom_type_l
df['srd_powierzchnia'] = area_l
df['srd_dlugosc'] = length_l
#df['srd_zlozonosc_lini'] = 

```

```python
df.to_csv("BDOT_10k_powiat_krotoszyński.csv", index=False, sep=";", encoding="utf-8")
```
